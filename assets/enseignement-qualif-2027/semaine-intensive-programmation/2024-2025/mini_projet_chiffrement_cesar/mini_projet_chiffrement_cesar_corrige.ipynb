```python
def chiffrement_cesar(message, decalage, type):
    if type == "D":
        decalage *= -1

    debut_alphabet_ascii = ord('A')
    nouveau_message = ""

    for caractere in message:
        if caractere.isalpha():
            
            nouveau_message += chr((ord(caractere) - debut_alphabet_ascii + decalage) % 26 + debut_alphabet_ascii)
        else:
            nouveau_message += caractere

    return nouveau_message
```

```python
print(chiffrement_cesar("BONJOUR", 4, "C"))
```
