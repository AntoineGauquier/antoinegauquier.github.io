```python
from random import randint
```

```python
is_playing = True
print("Bienvenue dans la roulette Python !")
solde = int(input("Quelle est votre mise de départ ?"))

while is_playing:
    print("Votre solde actuel est de " + str(solde) + ".")
    nombre_choisi = input("Sur quel nombre souhaitez vous parier ? (entre 1 et 36 tous deux inclus)")
    somme_pariee = int(input("Quel montant de votre solde actuel souhaitez-vous parier ? "))

    if somme_pariee > solde:
        print("Vous n'avez pas assez pour parier autant ... Vous pariez donc votre solde restant, soit " + str(solde) + ".")
        somme_pariee = solde

    print("Lancement de la balle ...")
    nombre_tombe = randint(0, 36)
    print("La balle tombe sur le " + str(nombre_tombe) + " !")
    if nombre_tombe == 0:
        solde -=  somme_pariee
        print("C'est la banque qui rafle la mise ! Votre nouveau solde est de " + str(solde) + ".")
    elif nombre_tombe == nombre_choisi:
        solde += somme_pariee * 36
        print("C'est gagné !! Votre nouveu solde est de " + str(solde) + ".")
    else:
        solde -=  somme_pariee
        print("Mince, c'est perdu ! Vous perdez la somme pariée, votre nouvelle solde est donc " + str(solde) + ".")

    if solde == 0:
        print("Vous n'avez plus d'argent ... A bientôt !")
        is_playing = False
    else:
        joue_encore = input("Souhaitez vous encore jouer ? (tapez \"oui\" ou \"non\")")
        is_playing = (joue_encore != "non")
```

```python
def manche_roulette(nombre_choisi, nombre_tombe, mise):
    if nombre_tombe == 0:
        print("C'est la banque qui rafle la mise !")
        return -mise
    elif nombre_tombe == nombre_choisi:
        print("C'est gagné !!")
        return mise * 36
    else:
        print("Mince, c'est perdu !")
        return -mise
```

```python
def lancer_balle_roulette():
    return randint(0, 36)
```

```python
def partie_roulette():
    print("Bienvenue dans la roulette Python!")
    
    solde = int(input("Quelle est votre mise de départ ?"))
    print("Votre solde actuelle est de " + str(solde) + ".")
    
    nombre_choisi = int(input("Sur quel nombre souhaitez vous parier ? (entre 1 et 36 tous deux inclus)"))
    print("Vous pariez sur le nombre " + str(nombre_choisi) + " !")

    print("Lancement de la balle ...")
    nombre_tombe = lancer_balle_roulette()
    print("Et c'est le " + str(nombre_tombe) + " qui tombe !")

    gains = manche_roulette(nombre_choisi, nombre_tombe, solde)
    solde += gains

    print("Votre solde devient donc " + str(solde) + ".")
```

```python
partie_roulette()
```
