```python
## TP 12

def gauche(i):
    return 2*i+1

def droite(i):
    return 2*i+2
```

```python
## TP 13
import math

def extraire_min(A):
    if len(A) == 0:
        return None

    minimum = A[0]
    dernier = A.pop()

    if len(A) > 0:
        A[0] = dernier
        entasser_min_dijkstra(A, 0)

    return minimum
```

```python
def entasser_min_dijkstra(A, i):
    l = gauche(i)
    r = droite(i)
    
    if l < len(A) and A[l][0] < A[i][0]:
        min = l
    else:
        min = i

    if r < len(A) and A[r][0] < A[min][0]:
        min = r

    if min != i:
        A[i], A[min] = A[min], A[i]
        A = entasser_min_dijkstra(A, min)
    return A

def construction_tas_min_dijkstra(A):
    for i in range(math.floor(len(A)/2) - 1, -1, -1):
        A = entasser_min_dijkstra(A, i)
    return A

def parent(i):
    return (i - 1) // 2
```

```python
def diminuer_cle(A, i):
    while i > 0 and A[parent(i)][0] > A[i][0]:
        p = parent(i)
        A[i], A[p] = A[p], A[i]
        i = p
```

```python
def dijkstra(G, source):
    distances = {}
    parents = {}
    S = set()

    for v in G:
        distances[v] = math.inf
        parents[v] = None

    distances[source] = 0

    Q = []
    for v in G:
        Q.append((distances[v], v))

    Q = construction_tas_min_dijkstra(Q)

    while len(Q) > 0:
        dist_u, u = extraire_min(Q)
        S.add(u)

        for v, poids in G[u]:
            if v not in S and distances[v] > dist_u + poids:
                distances[v] = dist_u + poids
                parents[v] = u

                # Pas dans le pseudo-code de l'algorithme dans le cours. 
                # En effet, on présume dans le cours que lorsque l'on altère la valeur de v.d, on met à jour la structure Q 
                # qui est un Tas-Min: le changement de valeur de v.d peut rendre le Tas invalide.
                # Ce n'est pas quelque-chose d'automatique dans notre implémentation, et il faut donc s'en assurer. 
                for i in range(len(Q)):
                    if Q[i][1] == v:
                        Q[i] = (distances[v], v)
                        diminuer_cle(Q, i)
                        break

    return distances, parents
```

```python
G = {
    'A': [('B', 1), ('C', 4)],
    'B': [('C', 2), ('D', 5)],
    'C': [('D', 1)],
    'D': []
}
```

```python
d, pi = dijkstra(G, 'A')
```
