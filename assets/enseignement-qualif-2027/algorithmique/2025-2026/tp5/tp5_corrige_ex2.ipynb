```python
def suite_fibonacci_iterative(n):
    if n == 0:
        return 0 
    elif n == 1:
        return 1

    F_n_moins_1 = 1
    F_n_moins_2 = 0
    
    i = 2
    while i <= n:
        F_n = F_n_moins_1 + F_n_moins_2
        F_n_moins_2 = F_n_moins_1
        F_n_moins_1 = F_n
        i += 1

    return F_n
```

```python
for i in range(16):
    print(f"F_{i} = {suite_fibonacci_iterative(i)}")
```

```python
def suite_fibonacci_recursive(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1

    return suite_fibonacci_recursive(n-1) + suite_fibonacci_recursive(n-2)
```

```python
for i in range(16):
    print(f"F_{i} = {suite_fibonacci_recursive(i)}")
```
