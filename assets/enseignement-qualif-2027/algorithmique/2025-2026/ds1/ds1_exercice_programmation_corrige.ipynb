```python
def deuxieme_plus_grand(liste):
    if len(liste) == 0:
        return
    
    max = float('-inf')
    deuxieme_max = float('-inf')

    for valeur in liste:
        if valeur > max:
            deuxieme_max = max
            max = valeur
        elif valeur > deuxieme_max:
            deuxieme_max = valeur
    
    return deuxieme_max
```

```python
def assia(l):
    while l:
        triee = []
        x = l[0]
        for i in range(len(l) - 1):
            if x > l[i + 1]:
                x = l[i+1]
            else:
                x = l[i]
        triee.append(x)
        l.remove(x)
    return triee[-1]
```

```python
print(assia([9, 8, 2, 1, 1, 9, 3, 5]))
print(assia([8, 9, 8, 1]))
print(assia([1, 7, 5, 3, 6, 2, 4, 9, 9, 7]))
```

```python
from random import randint

def generer_liste(taille):
    return [randint(0, 1000000) for _ in range(taille)]
```

```python
from time import time
from tqdm import tqdm

def mesurer_temps(n):
    temps_mesures = []
    tailles_listes = [taille for taille in range(0, 250 * n + 1, 250)]
    
    for taille in tqdm(tailles_listes):
        liste = generer_liste(taille)
        
        start = time()
        _ = deuxieme_plus_grand(liste)
        temps_mesures.append(time() - start)

    return temps_mesures
```

```python
tps = mesurer_temps(800)
```
