```python
def binomial_dynamique_recursive(n, k, memo):
    if k == 0 or n == k:
        return 1
    elif (n, k) in memo:
        return memo[(n, k)]

    memo[(n, k)] = binomial_dynamique_recursive(n-1, k-1, memo) + binomial_dynamique_recursive(n-1, k, memo)
    return memo[(n, k)]
```

```python
def afficher_triangle_pascal(n):
    memo = {}
    for i in range(n + 1):
        for j in range(i + 1):
            print(binomial_dynamique_recursive(i, j, memo), end=" ")
        print()
```

```python
afficher_triangle_pascal(5)
```
