## Code du TP7 (classes Sommet et Graphe)

```python
class Sommet:
    def __init__(self, id, label):
        self.id = id
        self.label = label
        self.voisins = []

    def ajout_voisin(self, nouveau_voisin):
        if nouveau_voisin not in self.voisins:
            self.voisins.append(nouveau_voisin)

class Graphe:
    def __init__(self):
        self.sommets = {}
        self.n = 0
        self.m = 0

    def nombre_sommets(self):
        return self.n

    def nombre_aretes(self):
        return self.m
        
    def ajout_sommet(self, classe_sommet, label = ""):
        nouveau_sommet = classe_sommet(self.n, label)
        self.sommets[self.n] = nouveau_sommet
        self.n += 1

    def ajout_arete(self, id_sommet_1, id_sommet_2):
        self.sommets[id_sommet_1].ajout_voisin(id_sommet_2)
        self.sommets[id_sommet_2].ajout_voisin(id_sommet_1)
        self.m += 1
    
    def voisinage_sommet(self, id_sommet):
        return self.sommets[id_sommet].voisins

    def __str__(self):
        if self.n == 0:
            return "Le graphe est vide."
        
        resultat = "Le graphe est composé des sommets :\n"
        
        for id in self.sommets:
            s = self.sommets[id]
            resultat += "-\tNuméro " + str(s.id)
            if s.label != "":
                resultat += " (label: " + s.label + ")"
            resultat += ", ayant pour voisin(s) le(s) sommet(s) numéro(s) : " + str(s.voisins) + "\n"
        resultat += "Au total, " + str(self.n) + " sommets et " + str(self.m) + " arêtes."
        return resultat
    
    def __getitem__(self, id):
        return self.sommets[id]
```

## Recherche en largeur (BFS)

```python
import math

class SommetBFS(Sommet):
    def __init__(self, id, label):
        Sommet.__init__(self, id, label)
        self.couleur = "blanc"
        self.distance = math.inf 
        self.parent = None

    def initialisation_bfs(self):
        self.couleur = "blanc"
        self.distance = math.inf
        self.parent = None
```

```python
from collections import deque

def parcours_en_largeur(graphe, id_origine):

    for id_sommet in graphe.sommets:
        graphe[id_sommet].initialisation_bfs()
    
    graphe[id_origine].couleur = "gris"
    graphe[id_origine].distance = 0
    graphe[id_origine].parent = None

    file = deque()
    file.appendleft(id_origine)

    nb_iterations = 1 # Uniquement dans le but de faire de l'affichage : inutile pour la recherche en soi.
    
    while file: #Utilise la méthode native bool()
        id_sommet_defile = file.pop()

        voisins_ajoutes = [] # Uniquement dans le but de faire de l'affichage : inutile pour la recherche en soi.
        
        for id_voisin in graphe.voisinage_sommet(id_sommet_defile):
            voisin = graphe[id_voisin]
            if voisin.couleur == "blanc":
                voisins_ajoutes.append(id_voisin) # Uniquement dans le but de faire de l'affichage : inutile pour la recherche en soi.
                
                voisin.couleur = "gris"
                voisin.distance = graphe[id_sommet_defile].distance + 1
                voisin.parent = id_sommet_defile
                
                file.appendleft(id_voisin)
                
        graphe[id_sommet_defile].couleur = "noir"
        print("Itération " + str(nb_iterations) + ". Sommet numéro " + str(id_sommet_defile) + " exploré. Sommets ajoutés à la file : " + str(voisins_ajoutes) + ". Etat de la file : " + str(file))
        nb_iterations += 1
```

```python
G = Graphe()
print("Nombre de sommets de G: " + str(G.nombre_sommets()))
print("Nombre d'arêtes de G: " + str(G.nombre_aretes()))
print(G)

print("\n========== Ajout de 4 sommets ==========\n")

for i in range(4):
    G.ajout_sommet(SommetBFS, "S_" + str(i))
print(G)

print("\n========== Ajout d'une arête par paire de sommet possible ==========\n")

for i in range(4):
    for j in range(i + 1, 4):
        G.ajout_arete(i, j)

print(G)
```

```python
parcours_en_largeur(G, 0)
```

```python
import random

def generer_graphe_aleatoire_bfs():
    G_aleat = Graphe()
    
    nb_sommets_min = 8
    nb_sommets_max = 15
    proba_creation_arete = 0.8
    
    nb_sommets_aleat = random.randint(nb_sommets_min, nb_sommets_max)
    
    for i in range(nb_sommets_aleat):
        G_aleat.ajout_sommet(SommetBFS)
    
    for i in range(nb_sommets_aleat):
        for j in range(nb_sommets_aleat):
            valeur_tiree = random.random()
            if valeur_tiree >= proba_creation_arete:
                G_aleat.ajout_arete(i, j)
    
    return G_aleat

G_aleat = generer_graphe_aleatoire_bfs()
print(G_aleat)
```

```python
parcours_en_largeur(G_aleat, 0)
```

## Parcours en profondeur (DFS)

```python
class SommetDFS(Sommet):
    def __init__(self, id, label):
        Sommet.__init__(self, id, label)
        self.couleur = "blanc"
        self.distance = math.inf 
        self.parent = None
        self.d = -1
        self.f = -1

    def initialisation_dfs(self):
        self.couleur = "blanc"
        self.distance = math.inf
        self.parent = None
        self.d = -1
        self.f = -1 
```

```python
def parcours_en_profondeur(graphe):
    for id_sommet in graphe.sommets:
        graphe[id_sommet].initialisation_dfs()
    temps = 0
    nb_iterations = 0
    
    for id_sommet in graphe.sommets:
        if graphe[id_sommet].couleur == "blanc":
            temps, nb_iterations = parcours_en_profondeur_sommet(graphe, id_sommet, temps, nb_iterations)

def parcours_en_profondeur_sommet(graphe, id_sommet, temps, nb_iterations):
    nb_iterations += 1
    print("Itération " + str(nb_iterations) + ". Sommet numéro " + str(id_sommet) + " exploré.")
    graphe[id_sommet].couleur = "gris"
    graphe[id_sommet].d = temps
    temps += 1

    for id_voisin in graphe.voisinage_sommet(id_sommet):
        voisin = graphe[id_voisin]
        if voisin.couleur == "blanc":
            voisin.parent = id_sommet
            temps, nb_iterations = parcours_en_profondeur_sommet(graphe, id_voisin, temps, nb_iterations)

    temps += 1
    graphe[id_sommet].couleur = "noir"
    graphe[id_sommet].f = temps
    return temps, nb_iterations
```

```python
G = Graphe()
print("Nombre de sommets de G: " + str(G.nombre_sommets()))
print("Nombre d'arêtes de G: " + str(G.nombre_aretes()))
print(G)

print("\n========== Ajout de 4 sommets ==========\n")

for i in range(4):
    G.ajout_sommet(SommetDFS, "S_" + str(i))
print(G)

print("\n========== Ajout d'une arête par paire de sommet possible ==========\n")

for i in range(4):
    for j in range(i + 1, 4):
        G.ajout_arete(i, j)

print(G)
```

```python
parcours_en_profondeur(G)
```
