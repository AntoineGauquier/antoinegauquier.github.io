```python
from time import time, time_ns, perf_counter
import matplotlib.pyplot as plt
import numpy as np
from tqdm import tqdm
```

```python
def sommation_entiers(n):
    k = 0
    for i in range(n):
        for j in range(i):
            k += 1
    
    return k
```

```python
def sommation_entiers_efficace(n):
    k = 0
    for i in range(n):
        k += i
        
    return k
```

```python
def sommation_entiers_directe(n):
    return n*(n + 1)/2
```

```python
valeurs_n = [i for i in range(2000)]
temps_mesures_sommation_entiers_directe = []
for n in tqdm(valeurs_n):
    temps_depart = perf_counter()
    somme = sommation_entiers_directe(n)
    temps_total = perf_counter() - temps_depart
    temps_mesures_sommation_entiers_directe.append(temps_total)
```

```python
temps_mesures_sommation_entiers_efficace = []
for n in tqdm(valeurs_n):
    temps_depart = perf_counter()
    somme = sommation_entiers_efficace(n)
    temps_total = perf_counter() - temps_depart
    temps_mesures_sommation_entiers_efficace.append(temps_total)
```

```python
temps_mesures_sommation_entiers = []
for n in tqdm(valeurs_n):
    temps_depart = perf_counter()
    somme = sommation_entiers(n)
    temps_total = perf_counter() - temps_depart
    temps_mesures_sommation_entiers.append(temps_total)
```

```python
complexite_asymptotique_theorique_quadratique = [(5 * 10 ** (-8)) * valeur ** 2 for valeur in valeurs_n]
complexite_asymptotique_theorique_lineaire = [(8 * 10 ** (-8)) * valeur for valeur in valeurs_n]

fig, ax = plt.subplots(figsize=(15, 10))
ax.plot(valeurs_n, temps_mesures_sommation_entiers, label="Temps mesuré empiriquement sommation_entiers")
ax.plot(valeurs_n, temps_mesures_sommation_entiers_efficace, label="Temps mesuré empiriquement sommation_entiers_efficace")
ax.plot(valeurs_n, temps_mesures_sommation_entiers_directe, label="Temps mesuré empiriquement sommation_entiers_directe")
ax.plot(valeurs_n, complexite_asymptotique_theorique_quadratique, label="Temps correspondant à une des fonction de O(n^2)")
ax.plot(valeurs_n, complexite_asymptotique_theorique_lineaire, label="Temps correspondant à une des fonction de O(n)")

ax.set(xlabel='Taille de l\'entrée n', ylabel='Temps (s)', title="Visualisation des complexités")
ax.grid()
plt.legend()

plt.show()
```
