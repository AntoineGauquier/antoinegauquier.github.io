```python
class Sommet:
    def __init__(self, id, label):
        self.id = id
        self.label = label
        self.voisins = []

    def ajout_voisin(self, nouveau_voisin):
        if nouveau_voisin not in self.voisins:
            self.voisins.append(nouveau_voisin)

class Graphe:
    def __init__(self):
        self.sommets = {}
        self.n = 0
        self.m = 0

    def nombre_sommets(self):
        return self.n

    def nombre_aretes(self):
        return self.m
        
    def ajout_sommet(self, classe_sommet, label = ""):
        nouveau_sommet = classe_sommet(self.n, label)
        self.sommets[self.n] = nouveau_sommet
        self.n += 1

    def ajout_arete(self, id_sommet_1, id_sommet_2):
        self.sommets[id_sommet_1].ajout_voisin(id_sommet_2)
        self.sommets[id_sommet_2].ajout_voisin(id_sommet_1)
        self.m += 2 

    def ajout_arete_ponderee(self, id_sommet_1, id_sommet_2, poids):
        self.sommets[id_sommet_1].ajout_voisin({'id':id_sommet_2, 'poids':poids})
        self.sommets[id_sommet_2].ajout_voisin({'id':id_sommet_1, 'poids':poids})
        self.m += 2

    def ajout_arete_digraphe(self, id_sommet_1, id_sommet_2, arete_de_1_vers_2, arete_de_2_vers_1):
        if arete_de_1_vers_2:
            self.sommets[id_sommet_1].ajout_voisin(id_sommet_2)
            self.m += 1
        if arete_de_2_vers_1:
            self.sommets[id_sommet_2].ajout_voisin(id_sommet_1)
            self.m += 1
    
    def voisinage_sommet(self, id_sommet):
        return [id_voisin for id_voisin in self.sommets[id_sommet].voisins]

    def __str__(self):
        if self.n == 0:
            return "Le graphe est vide."
        
        resultat = "Le graphe est composé des sommets :\n"
        
        for id in self.sommets:
            s = self.sommets[id]
            resultat += "-\tNuméro " + str(s.id)
            if s.label != "":
                resultat += "(label: " + s.label + ")"
            resultat += ", ayant pour voisin(s) le(s) sommet(s) numéro(s) : " + str(s.voisins) + "\n"
        resultat += "Au total, " + str(self.n) + " sommets et " + str(self.m) + " arêtes."
        return resultat
    
    def __getitem__(self, id):
        return self.sommets[id]
```

```python
class Ensemble:
    def __init__(self, parent):
        self.parent = parent
        self.rang = 0

def creer_ensemble(x):
    return Ensemble(x)

def trouver(ensembles, x):
    if x != ensembles[x].parent:
        ensembles, ensembles[x].parent = trouver(ensembles, ensembles[x].parent)
    return ensembles, ensembles[x].parent

def lier(ensembles, x, y):
    if x == y:
        return
    if ensembles[x].rang > ensembles[y].rang:
        ensembles[y].parent = x
    else:
        ensembles[x].parent = y
        if ensembles[x].rang == ensembles[y].rang:
            ensembles[y].rang += 1
    return ensembles

def union(ensembles, x, y):
    ensembles, trouver_x = trouver(ensembles, x)
    ensembles, trouver_y = trouver(ensembles, y)
    return lier(ensembles, trouver_x, trouver_y)
```

```python
class AreteKruskal:
    def __init__(self, id_sommet_1, id_sommet_2, poids):
        self.id_sommet_1 = id_sommet_1
        self.id_sommet_2 = id_sommet_2
        self.poids = poids

def algorithme_kruskal(graphe):
    A = [] # Liste des objets AreteKruskal à retourner une fois l'ACPM calculé
    
    # Création d'une liste d'objets Ensemble (un objet par sommet dans graphe)
    objets_ensemble = []
    for sommet in graphe.sommets:
        objets_ensemble.append(Ensemble(sommet))

    aretes = []
    for sommet in graphe.sommets: # On utilise la liste d'adjacence pour créer notre liste d'arêtes "AreteKruskal"
        for voisin in graphe[sommet].voisins:
            aretes.append(AreteKruskal(sommet, voisin['id'], voisin['poids']))

    aretes.sort(key =lambda arete:arete.poids) # On trie la liste de toutes les arêtes par ordre croissant de poids (fait appel aux lambdas Python)
    
    # Reste de l'algorithme Kruskal ... Devant faire appel aux fonctions d'Union-Find.
    for arete in aretes:
        objets_ensemble, trouver_sommet_1 = trouver(objets_ensemble, arete.id_sommet_1)
        objets_ensemble, trouver_sommet_2 = trouver(objets_ensemble, arete.id_sommet_2)
        
        if trouver_sommet_1 != trouver_sommet_2:
            A.append(arete)
            objets_ensemble = union(objets_ensemble, arete.id_sommet_1, arete.id_sommet_2)
    return A
```
