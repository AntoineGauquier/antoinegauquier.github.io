```python
from collections import deque

def deux_coloriable(V):
    file = deque()
    couleurs = [-1 for _ in range(len(V))]

    couleurs[0] = 0
    file.appendleft(V[0])

    while file:
        sommet = file.pop()

        for voisin in liste_adjacence(sommet):
            if couleurs[voisin] == -1:
                couleurs[voisin] = 1 - couleurs[sommet]
                file.append(voisin)
            else:
                if couleurs[voisin] == couleurs[sommet]:
                    return False
    return True
```

```python
# Graph Structure:
# 0 - 1
# |   |
# 3 - 2

V = [i for i in range(4)]
adj = [[] for _ in range(len(V))]

adj[0].append(1)
adj[1].append(0)
adj[1].append(2)
adj[2].append(1)
adj[2].append(3)
adj[3].append(2)
adj[3].append(0)
adj[0].append(3)

def liste_adjacence(sommet):
    global adj
    return adj[sommet]
```

```python
from collections import deque

def a(V):
    c  = []
    for u in V:
        c.append("rouge")
    s = 0
    c[s] = "bleu"
    print(c)
    Q = deque()
    Q.appendleft(0)
    while Q:
        print(Q)
        u = Q.pop()
        for v in liste_adjacence(u):
            if c[v] == c[u]:
                return False

            if c[v] == "rouge":
                c[v] = "bleu"
                Q.appendleft(v)
                print(c)
    return True
```

```python
from time import sleep
def b(V):
    file = deque()
    file.appendleft(V[0])
    while file:
        s = file.pop()
        print(file)
        sleep(0.5)
        vt = []
        for v in liste_adjacence(s):
            for u in liste_adjacence(v):
                vt.append(u)
            if v in vt:
                return False
            file.appendleft(v)
    return True
```

```python
def d(V):
    c = [1 for _ in range(len(V))]
    for s in V:
        for v in liste_adjacence(s):
            c[v] = -c[s]
    for s in V:
        for v in liste_adjacence(s):
            if c[v] == c[s]:
                return False
    return True
```
