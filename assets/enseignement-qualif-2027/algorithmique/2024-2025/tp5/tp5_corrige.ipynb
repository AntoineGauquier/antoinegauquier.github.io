# Suite de Fibonacci

```python
def suite_fibonacci_dynamique(n):
    valeurs_fibonacci = [None for _ in range(n + 1)]
    valeurs_fibonacci[0] = 0
    if n > 0:
        valeurs_fibonacci[1] = 1
    
        for i in range(2, n + 1):
            valeurs_fibonacci[i] = valeurs_fibonacci[i - 1] + valeurs_fibonacci[i - 2]
    
    return valeurs_fibonacci[-1]
```

```python
for i in range(20):
    print("F_" + str(i) + " = " + str(suite_fibonacci_dynamique(i)))
```

# Rendu de monnaie (nombre minimal de pièces)

```python
import math
import sys

sys.setrecursionlimit(100000)
```

```python
def nombre_minimal_pieces_memoisation(P, S, k, dict_memoisation):
    cle = (S, k)

    if cle in dict_memoisation:
        return dict_memoisation[cle]

    if S == 0:
        dict_memoisation[cle] = 0
    elif (k == -1 and S >= 0) or S < 0:
        dict_memoisation[cle] = math.inf
    else:
        dict_memoisation[cle] = min(nombre_minimal_pieces_memoisation(P, S, k - 1, dict_memoisation), nombre_minimal_pieces_memoisation(P, S - P[k], k, dict_memoisation) + 1)
    
    return dict_memoisation[cle]
```

```python
P_euro = [1, 2, 5, 10, 20, 50, 100, 200]
S_1 = 1301
print("Le nombre de pièces à rendre pour S = " + str(0) + " avec le système de pièces " + str(P_euro) + " est " + str(nombre_minimal_pieces_memoisation(P_euro, 0, len(P_euro) - 1, dict())))
print("Le nombre de pièces à rendre pour S = " + str(0) + " avec le système de pièces " + str(P_euro) + " est " + str(nombre_minimal_pieces_memoisation(P_euro, 2, -1, dict())))

print("Le nombre de pièces à rendre pour S = " + str(S_1) + " avec le système de pièces " + str(P_euro) + " est " + str(nombre_minimal_pieces_memoisation(P_euro, S_1, len(P_euro) - 1, dict())))

P_petites_coupures = [1, 2, 5]
S_2 = 113

print("Le nombre de pièces à rendre pour S = " + str(S_2) + " avec le système de pièces " + str(P_petites_coupures) + " est " + str(nombre_minimal_pieces_memoisation(P_euro, S_2, len(P_petites_coupures) - 1, dict())))
```

# Problème du sac à dos

```python
def sac_a_dos_dynamique(P, V, C):
    struct_dynamique = [[None for _ in range(len(P) + 1)] for _ in range(C + 1)]

    for k in range(len(P) + 1):
        for c in range(C + 1):
            if k == 0 or c == 0:
                val = 0
            elif P[k - 1] > c:
                val = struct_dynamique[c][k - 1]
            else:
                val = max(V[k - 1] + struct_dynamique[c - P[k - 1]][k - 1], struct_dynamique[c][k - 1])

            struct_dynamique[c][k] = val

    return struct_dynamique[C][len(P)]
```

```python
def sac_a_dos_recursive(P, V, C, k=None):
    if k is None:
        k = len(P)

    if k == 0 or C == 0:
        return 0
    
    if P[k-1] > C:
        return sac_a_dos_recursive(P, V, C, k-1)
    
    return max(sac_a_dos_recursive(P, V, C, k-1), V[k-1] + sac_a_dos_recursive(P, V, C - P[k-1], k-1))
```

```python
P = [10, 4, 4]
V = [70, 40, 40]
C = 8

sac_a_dos_dynamique(P, V, C)
```

```python
P = [10, 4, 4]
V = [70, 40, 40]
C = 8

sac_a_dos_recursive(P, V, C)
```
