```python
from time import time
from tqdm import tqdm
from random import randint
import matplotlib.pyplot as plt
```

```python
def indice_minimum(A):
    if len(A) == 0:
        return None
    
    indice_minimum = 0
    minimum = A[0]
    
    for i in range(1, len(A)):
        if A[i] < minimum:
            minimum = A[i]
            indice_minimum = i
    
    return indice_minimum
```

```python
def mesurer_temps(n):
    tailles = [100*i for i in range(0,n+1)]
    temps = []

    for taille in tqdm(tailles):
        tab = generer_tableau(taille)
        print(tab)
        debut = time()
        indice_minimum(tab)
        temps.append(time() - debut)

    return tailles, temps
```

```python
def generer_tableau(n):
    return[randint(0, n) for _ in range(n)]
```

```python
tailles, temps = mesurer_temps(10)
```

```python
complexite_asymptotique_theorique = [(0.15 * 10 ** (-7)) * valeur for valeur in tailles]

fig, ax = plt.subplots(figsize=(15, 10))
ax.plot(tailles, temps, label="Temps mesuré empiriquement")
ax.plot(tailles, complexite_asymptotique_theorique, label="Complexité théorique (logarithmique)")
ax.set(xlabel='Taille de l\'entrée n', ylabel='Temps (s)', title="Visualisation des complexités")
#ax.set_yscale('log')
ax.grid()
plt.legend()

plt.show()
```
