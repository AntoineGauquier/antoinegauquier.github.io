```python
def binomial_memoisation(n, k, struct_memo):
    key = (n, k)

    if key in struct_memo:
        return struct_memo[key]

    if k == 0 or k == n:
        return 1

    struct_memo[key] = binomial_memoisation(n - 1, k - 1, struct_memo) + binomial_memoisation(n - 1, k, struct_memo)
    return struct_memo[key]
```

```python
def binomial_sans_memoisation(n, k):
    struct_dynamique = [[None for _ in range(k + 1)] for _ in range(n + 1)]

    for i in range(n + 1):
        for j in range(min(i, k) + 1):
            if j == 0 or j == i:
                struct_dynamique[i][j] = 1
            else:
                struct_dynamique[i][j] = struct_dynamique[i - 1][j - 1] + struct_dynamique[i - 1][j]

    return struct_dynamique[n][k]
```

```python
val = 24

for i in range(val):
    tp = ""
    for j in range(min(i, val) + 1):
        tp += str(binomial_sans_memoisation(i, j)) + " "
    print(tp)
```

Le plus simple est de regarder la version sans mémoïsation car le procédé est itératif. 

Pour la complexité temporelle, on a deux boucles imbriquées, la première faisant $(n+1)$ itérations, et la seconde au maximum $k$, ce qui fait au total un nombre d'itérations en $O\left(n\times k\right)$. Toutes les lignes au sein de ces boucles ou à l'extérieur de celles-ci se font en temps constant (accès à des listes, sommes, retour), excepté pour la première ligne d'initialisation de la structure qui est en $O\left(\left(n + 1\right) \times \left(k + 1\right)\right) = O\left(n\times k\right)$. Pour synthétiser, cela fait du $O\left(n\times k\right)$.

Concernant maintenant la complexité spatiale, on créé un tableau en deux dimensions à $(n+1)$ lignes et $(k+1)$ colonnes, donc on créé, en plus des variables passées en paramètre, une structure de données de taille $(n+1)\times(k+1)$. D'autres variables sont crées (comme $i$ et $j$) mais ne stockent qu'une unique valeur, indépendemment des paramètres $n$ et $k$. Ainsi, la complexité asymptotique spatiale de cet algorithme est en $O\left(n\times k\right)$.
